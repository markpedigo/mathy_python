## A first function

```python
def circle_area(radius):
    area = 3.14159 * radius ** 2
    return area
```

```python
circle_area(7)
```

```python
circle_area(3)
```

```python
circle_area(10)
```

```python
a = circle_area(7)
```

## Encapsulation

```python
import math

def circle_area(radius):
    area = math.pi * radius ** 2
    return area
```

```python
def classify_temp(temp_f):
    if temp_f <= 32:
        return 'freezing'
    if temp_f < 70:
        return 'mild'
    return 'warm'
```

## Parameters and Arguments

```python
def circle_area(radius):  # radius is a parameter
    area = 3.14159 * radius ** 2
    return area
```

```python
circle_area(7)  # 7 is an argument = the value of a parameter
```

## Multiple parameters and keyword arguments

```python
def compound_interest(principal, rate, years):
    return principal * (1 + rate) ** years
```

```python
compound_interest(1000, 0.05, 10)
```

```python
compound_interest(years=10, principal=1000, rate=0.05)
```

```python
compound_interest(1000, years=10, rate=0.05)
```

```python
compound_interest(years=10, rate=0.05, 1000)
```

## Functions calling other functions

```python
def celsius_to_fahrenheit(c):
    return c * 9/5 + 32

def describe_temp(c):
    f = celsius_to_fahrenheit(c)
    if f <= 32:
        return f'{f:.1f}°F is freezing.'
    elif f < 70:
        return f'{f:.1f}°F is cool.'
    else:
        return f'{f:.1f}°F is warm.'
```

```python
describe_temp(21)
```
