3. Write a function `celsius_to_fahrenheit(c)` and a separate function `fahrenheit_to_celsius(f)`. Verify they are inverses of each other by converting 100°C to Fahrenheit and back.

````python
def celsius_to_fahrenheit(c):
    f = (c * 1.8) + 32
    return f

def fahrenheit_to_celsius(f):
    c = (f - 32) * 5 / 9
    return c

f = celsius_to_fahrenheit(100)
print(f)

c = fahrenheit_to_celsius(f)
print(c)
````

4. Write a function `circle_stats(radius)` that returns both the area and circumference of a circle. Test it on radius 5.

````python
import math

def circle_stats(radius):
    area = round(math.pi * radius**2, 1)
    circumference = round(2 * math.pi * radius, 1)
    return area, circumference

circle_stats(5)
````

7. Write a function `summarize(numbers)` that takes a list and returns a dictionary with keys `'min'`, `'max'`, `'sum'`, `'count'`, and `'mean'`. Round each stat to two decimal places.

````python
def summarize(numbers):
    
    min = 9999
    max = -9999
    total = 0
    count = 0
    for number in numbers:
        if number < min:
            min = number
        if number > max:
            max = number
        total += number
        count += 1
    mean = total / count

    result = {
        'min': round(min, 2),
        'max': round(max, 2),
        'sum': round(total, 2),
        'count': count,  # always an int, don't need to round
        'mean': round(mean, 2)
    }
    return result

numbers = [6, 5, 7.6, 992, -34, 45, 5, 3.2]
summarize(numbers)
````

8. Write a function `is_palindrome(s)` that returns `True` if a string is a palindrome, ignoring spaces and case. Test it on `'racecar'`, `'A man a plan a canal Panama'`, and `'Mathy Python'`.

````python
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

print(is_palindrome('racecar'))
print(is_palindrome('A man a plan a canal Panama'))
print(is_palindrome('Mathy Python'))
````
