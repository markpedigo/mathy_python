11. **Prime numbers.** Write a function `is_prime(n)` that returns `True` if `n` is prime and `False` otherwise. Use it to print all primes up to 100. How many are there?

````python
import math

def is_prime(n):
    for i in range(2, int(math.sqrt(n) + 1)):
        if n % i == 0:
            return False
    return True

count = 0
primes = []
for i in range(2, 101):
    if is_prime(i):
        count += 1
        primes.append(i)
print(primes)
print(count)
````

12. **Run-length encoding.** Given a string like `'aaabbbccddddee'`, compress it to `'a3b3c2d4e2'`, which is each character followed by its count. Then write the decoder that reverses the process.

````python
def encode(s):
    prev = ''
    result = ''
    count = 1
    for char in s:
        if char == prev:
            count += 1
        else:
            if count > 1:
                result += str(count)
            result += char
            count = 1
        prev = char
    # Could miss char on end, flush it
    if count > 1:
        result += str(count)
    return result

def decode(s):
    result = ''
    prev = ''
    for char in s:
        if char.isdigit():
            result += prev * (int(char) - 1)
        else:
            result += char
        prev = char
    return result

x = 'aaabbbccddddee'
x_encoded = encode(x)
print(x_encoded)
x_decoded = decode(x_encoded)
print(x_decoded)
````

13. **Sieve of Eratosthenes.** Implement the classic algorithm to find all primes up to N. (Google it if needed.)

````python
prime_indices = [True] * 101  # 0, 1, ..., 100

# Index = number, value = True/False
# So [False, False, True, True, ...] means 0 and 1 are not prime, 2 is prime, 3 is prime, ...
prime_indices[0] = False
prime_indices[1] = False
for i in range(2, int(math.sqrt(100)) + 1):
    if prime_indices[i]:
        for j in range(i*i, 101, i):
            prime_indices[j] = False

# "Unroll" the prime indices to thir corresponding numbers
primes = []
for i in range(2, len(prime_indices)):
    if prime_indices[i]:
        primes.append(i)
print(primes)
print(len(primes))
````

14. **Longest streak.** Given a list of daily temperatures:
```python
temps = [65, 68, 72, 71, 75, 78, 80, 79, 76, 70, 68, 72, 74, 77, 80]
```
Find the longest streak of consecutive days where the temperature increased each day. Print the streak length, the starting day index, and the starting temperature.

````python
temps = [65, 68, 72, 71, 75, 78, 80, 79, 76, 70, 68, 72, 74, 77, 80]

streak = 0
max_streak = 0
for i in range(1, len(temps)):
    diff = temps[i] - temps[i-1]
    if diff > 0:
        streak += 1
        if streak > max_streak:
            max_streak = streak
            starting_day = i - max_streak
            starting_temp = temps[starting_day]
    else:
        streak = 0

print(f'max_streak = {max_streak}')
print(f'starting_day = {starting_day}')
print(f'starting_temp = {starting_temp}')
````

15. **Monte Carlo pi.** Estimate π using the Monte Carlo method:
* Generate N random points in the unit square,
* Count how many fall inside the unit circle,
* $\pi$ is approximately 4 times the proportion of points inside the circle.

How many points do you need for 2 decimal places of accuracy?

Hint: Use `random.random()` for the point generation. Look up the `random` module [here](https://docs.python.org/3/library/random.html).

````python
import random

random.seed(42)  # So that our "random" results are consistent.
                 # 42 is a common value. It's a Hitchhiker's Guide to the Galaxy in-joke.

def monte_carlo_pi(N):
    # Generate N random points in the unit square
    # Use a list of tuples
    points = []
    for i in range(N):
        x = random.random()
        y = random.random()
        points.append((x,y))  # inner () to denote tuple

    # Count how many fall inside the unit circle
    count = 0
    for point in points:
        x = point[0]
        y = point[1]
        if x**2 + y**2 < 1:
            count += 1
            
    # Calculate ratio
    pi_approx = 4 * count / N
    return pi_approx

test_points = [1, 10, 100, 1000, 10000, 100000, 1000000]
print(f'{"N":<10} "pi"')
print('-' * 20)
for N in test_points:
    print(f'{N:<10,} {monte_carlo_pi(N)}')
````

You need 100,000 points for two decimal points of accuracy.
