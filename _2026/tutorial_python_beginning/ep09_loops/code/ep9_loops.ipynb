## for loops over a list

```python
scores = [88, 92, 75, 100, 84]

for score in scores:
    print(score)
```

```python
scores = [88, 92, 75, 100, 84]

for score in scores:
    print(round(score * 1.1))
```

## The accumulator pattern

```python
total = 0 
for score in scores:
    total += score

print(total)
```

# while loops

```python
balance = 1000
rate = 0.05
years = 0

while balance < 2000:
    balance *= (1 + rate)
    years += 1

print(f'Balance doubles after {years} years.')
```

```python
# infinite loop
balance = 1000
rate = 0.05
years = 0

while balance < 2000:
    balance *= (1 + rate)
    years += 1

print(f'Balance doubles after {years} years.')
```

## break and continue

```python
while True:
    response = input("Enter a score, or 'q' to quit")
    if response == 'q':
        break
    print(f'You entered {response}.')

print('You quit the program.')
```

## Looping over two things at once with enumerate and zip

```python
temps = [58, 61, 60, 72, 70, 68, 75]

i = 0
biggest_jump = 0
jump_day = 0

for temp in temps:
    if i > 0:
        jump = abs(temp - temps[i-1])
        if jump > biggest_jump:
            biggest_jump = jump
            jump_day = i
    i += 1

print(f'The biggest jump was {biggest_jump} degrees, on day {jump_day}.')
```

```python
temps = [58, 61, 60, 72, 70, 68, 75]

biggest_jump = 0
jump_day = 0

for i, temp in enumerate(temps[1:], start=1):
    jump = abs(temp - temps[i-1])
    if jump > biggest_jump:
        biggest_jump = jump
        jump_day = i

print(f'The biggest jump was {biggest_jump} degrees, on day {jump_day}.')
```

```python
names = ['Alice', 'Bob', 'Carol', 'Dave', 'Eve']
scores = [88, 92, 75, 100, 84]

for i in range(len(names)):
    if scores[i] > 90:
        print(names[i], scores[i])
```

```python
for name, score in zip(names, scores):
    if score > 90:
        print(name, score)
```
