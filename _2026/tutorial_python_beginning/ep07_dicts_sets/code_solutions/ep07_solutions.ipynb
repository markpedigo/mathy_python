9. Given two sets of students:

Find: students in both classes, students in math but not python, students in either class but not both, and total unique students across both classes.

````python
math_class = {'Alice', 'Bob', 'Carol', 'Dave', 'Eve'}
python_class = {'Bob', 'Dave', 'Frank', 'Grace'}

both_classes = math_class | python_class
print('Both:', both_classes)

either_class = len(math_class - python_class) + len(python_class - math_class)
print('Either:', either_class)

total_unique = len(math_class | python_class)
print('Total unique:', total_unique)
````

11. **Inverted index.** Given:
```python
grades = {'Alice': 'A', 'Bob': 'C', 'Carol': 'B', 'Dave': 'A', 'Eve': 'B'}
```
Build a new dictionary that maps each grade to a list of students who received it. Expected result: `{'A': ['Alice', 'Dave'], 'C': ['Bob'], 'B': ['Carol', 'Eve']}`.

````python
grades = {'Alice': 'A', 'Bob': 'C', 'Carol': 'B', 'Dave': 'A', 'Eve': 'B'}

reverse_lookup = {}
for student, grade in grades.items():
    if grade in reverse_lookup:
        reverse_lookup[grade].append(student)
    else:
        reverse_lookup[grade] = [student]

reverse_lookup
````

14. **Histogram.** Given a list of exam scores:
```python
scores = [88, 92, 75, 100, 84, 91, 67, 78, 95, 88, 73, 82]
```
Build a dictionary that maps grade bands ('A': 90-100, 'B': 80-89, 'C': 70-79, 'D': 60-69) to the count of scores in each band. Print the result as a simple text histogram:
```
A: ****
B: *****
C: ***
D: *
```

````python
scores = [88, 92, 75, 100, 84, 91, 67, 78, 95, 88, 73, 82]
grade_bands = {
    'A': 0,
    'B': 0,
    'C': 0,
    'D': 0,    
}

# calculate bands
for score in scores:
    if score >= 90:
        grade_bands['A'] += 1
    elif score >= 80:
        grade_bands['B'] += 1
    elif score >= 70:
        grade_bands['C'] += 1
    elif score >= 60:
        grade_bands['D'] += 1
    else:
        grade_bands['F'] += 1

# print out histogram
for grade in grade_bands:
    print(grade, '*' * grade_bands[grade])
````

15. **Word co-occurrence.** Given a list of sentences:
```python
sentences = [
    'the cat sat on the mat',
    'the cat ate the rat',
    'the rat sat on the mat'
]
```
Build a dictionary where each key is a word and the value is a set of all words that appear in the same sentence as that word. For example, `'cat'` should map to `{'the', 'sat', 'on', 'mat', 'ate', 'rat'}`.

````python
sentences = [
    'the cat sat on the mat',
    'the cat ate the rat',
    'the rat sat on the mat'
]

flat = set()
for sentence in sentences:
    words = sentence.split()
    flat.update(words)

for word in flat:
    for sentence in sentences:
        if word in sentence:
            print(word, "->", set(sentence.split()) - {word})
````
