## Iterating over dictionaries

```python
# Views into a dictionary
```

```python
instructor.keys()
```

```python
instructor.values()
```

```python
instructor.items()
```

```python
for key in instructor.keys():
    print(key)
```

```python
for value in instructor.values():
    print(value)
```

```python
for key, value in instructor.items():
    print(key, '->', value)
```

```python
'name' in instructor.keys()
```

```python
'Mark' in instructor.keys()
```

```python
'Mark' in instructor.values()
```

### Lookups go one direction in dicts

```python
instructor['name']
```

```python
target = 'Mark'
for key, value in instructor.items():
    if value ==  target:
        print(key)
```

## Sets

```python
primes = {2, 3, 5, 7, 11}
```

```python
primes_with_dups = {1, 2, 2, 1, 3, 3, 5, 7, 11, 1, 2}
```

```python
primes_with_dups
```

```python
scores = [88, 92, 75, 92, 88, 100]
```

```python
scores
```

```python
set(scores)
```

```python
scores_new = sorted(list(set(scores)))
scores_new
```

```python
primes = {2, 3, 5, 7, 11}
odds = {1, 3, 5, 7, 9, 11}
```

```python
primes & odds
```

```python
primes | odds
```

```python
primes - odds  # primes, not odds
```

```python
odds - primes  # odds, not primes
```

```python
primes ^ odds  # symmetric difference. primes or odds, but not in the intersection
```

## Choosing the right collection

* Need order and the ability to change elements? List
* Need order but the data shouldn't change? Tuple
* Need to look things up by name? Dictionary
* Need unique elements, or want to do set math? Set
