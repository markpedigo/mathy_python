4. Write a program that classifies a number as a perfect square, a perfect cube, both, or neither. Test it on 64, 27, 36, 100, and 11. Hint: Look in the `math` module [here](https://docs.python.org/3/library/math.html). 

````python
x = int(input('Enter a number:'))

square_root = math.sqrt(x)
is_square = math.isclose(square_root, int(square_root))
if is_square:
    
    print(f'Perfect square! Square root = {int(square_root)}.')
    
cube_root = math.cbrt(x)
is_cube = math.isclose(cube_root, int(cube_root))
if is_cube:
    print(f'Perfect cube! Cube root = {int(cube_root)}.')

if not (is_square and is_cube):
    print('Neither a square nor a cube.')
````

9. Write a password validator. A valid password must:
- Be at least 8 characters long
- Contain at least one uppercase letter
- Contain at least one digit

Print a specific message for each rule that fails. If all rules pass, print `'Password accepted.'`

````python
password = input('Enter a password')

is_valid = True

# Length >= 8 characters?
if len(password) < 8:
    is_valid = False
    print('Password must be at least 8 characters long.')

# Contains at least one upper case letter?
is_upper = False
for char in password:
    if char.isupper():
        is_upper = True
        break
# if not any(char.isupper() for char in password)  # Easier, but we haven't covered this construct yet
if not is_upper:
    is_valid = False
    print('Password must contain at least one uppercase letter.')

# Contains at least one digit?
is_digit = False
for char in password:
    if char.isdigit():
        is_digit = True
        break
if not is_digit:
    is_valid = False
    print('Password must contain at least one digit.')

if is_valid:
    print('Password accepted.')
````

12. **Triangle classifier.** Given three side lengths, determine:
- First, whether they can form a valid triangle (triangle inequality theorem)
- If valid, whether the triangle is equilateral, isosceles, or scalene
- If valid, whether it is acute, right, or obtuse (use the Pythagorean theorem)

````python
def is_valid_triangle(a, b, c):
    '''
    A triangle is valid if the sum of the lengths of any two sides is greater than the length of the third side
    '''
    return (a + b > c) and (b + c > a) and (a + c > b)

def classify_sides(a, b, c):
    # Assumes valid triangle.
    if a == b == c:  # legal Python. "comparison operator chaining"
        return 'Equilateral'
    elif (a == b) or (b == c) or (a == c):
        return 'Isosceles'
    else:
        return 'Scalene'

def classify_angles(a, b, c):
    # Assumes valid triangle.
    a, b, c = sorted([a, b, c])  # c needs to be the longest side
    
    if math.isclose(c**2, a**2 + b**2):
        return 'right triangle'
    elif a**2 + b**2 > c**2:
        return 'acute triangle'
    else:
        return 'obtuse triangle'

def classify_triangle(a, b, c):
    if is_valid_triangle(a, b, c):
        result = f'{classify_sides(a, b, c)}, {classify_angles(a, b, c)}.'
    else:
        result = 'Not a valid triangle.'
    return result

print(classify_triangle(3, 4, 5))     # Scalene, right triangle.
print(classify_triangle(5, 5, 5))     # Equilateral, acute triangle.
print(classify_triangle(5, 5, 8))     # Isosceles, obtuse triangle.
print(classify_triangle(1, 2, 10))    # Not a valid triangle.
print(classify_triangle(5, 12, 13))   # Scalene, right triangle.
````

13. **Collatz conjecture.** Starting from any positive integer n, apply this rule repeatedly: if n is even, divide by 2; if n is odd, multiply by 3 and add 1. The conjecture is that you always eventually reach 1. Write code that generates the sequence from a starting number and counts the steps. Which starting number under 100 takes the most steps?

````python
def collatz(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:  # even
            n //= 2
        else:
            n = n * 3 + 1
        steps += 1

        if steps > 1000:  # just in case; it's a conjecture not a theorem
            break
    return steps

max_steps_index = -999
for i in range(2, 101):
    if collatz(i) > max_steps_index:
        max_steps_index = i
print(f'{max_steps_index}, with {collatz(max_steps_index)} steps.')
````

14. **BMI calculator.** Given weight in pounds and height in inches, compute BMI using the formula `BMI = (weight / height²) * 703`. Classify the result as underweight (under 18.5), normal (18.5-24.9), overweight (25-29.9), or obese (30 and above). Print the BMI rounded to one decimal place and the classification.

````python
def classify_bmi(weight, height):
    bmi = round((weight / height **2 ) * 703, 1)
    print(f'BMI = {bmi}')

    if bmi < 18.5:
        classification = "underweight"
    elif bmi <= 24.9:
        classification =  "normal"
    elif bmi <= 29.9:
        classification = "overweight"
    else:
        classification = "obse"

    print(f'The patient has a BMI classification of {classification}.')
    print('-' * 100)

classify_bmi(120, 65)   # 5'5", 120 lbs -- underweight
classify_bmi(150, 68)   # 5'8", 150 lbs -- normal
classify_bmi(200, 68)   # 5'8", 200 lbs -- overweight
classify_bmi(250, 68)   # 5'8", 250 lbs -- obese
classify_bmi(130, 60)   # 5'0", 130 lbs -- boundary normal/overweight
````

15. **Number to words.** Given any integer from 0 to 999, print it in words. For example, 742 should print `'seven hundred forty two'`. Use dictionaries for the word lookups and conditionals for the logic.

````python
def number_to_words(x):    
    HUNDREDS_WORD = {
        0: '',
        1: 'one',
        2: 'two',
        3: 'three',
        4: 'four',
        5: 'five',
        6: 'six',
        7: 'seven',
        8: 'eight',
        9: 'nine'
    }
    
    TENS_WORD = {
        0: '',
        1: 'ten',
        2: 'twenty',
        3: 'thirty',
        4: 'forty',
        5: 'fifty',
        6: 'sixty',
        7: 'seventy',
        8: 'eighty',
        9: 'ninety'
    }

    TEENS_ONE_WORD = {
        1: 'eleven',
        2: 'twelve',
        3: 'thirteen',
        4: 'fourteen',
        5: 'fifteen',
        6: 'sixteen',
        7: 'seventeen',
        8: 'eighteen',
        9: 'nineteen'
    }

    ONES_WORD = HUNDREDS_WORD

    if x == 0:
        return 'zero'

    # Extract digits
    hundreds_digit = x // 100
    hundreds_remainder = x % 100
    tens_digit = hundreds_remainder // 10
    ones_digit = hundreds_remainder % 10

    # Build word
    result = ''
    if hundreds_digit != 0:
        result += HUNDREDS_WORD[hundreds_digit] + ' hundred '
    
    if tens_digit == 1:  # e.g., eleven
        result += TEENS_ONE_WORD[ones_digit] + ' '
        return result
    
    if tens_digit != 0:
        result += TENS_WORD[tens_digit] + ' '

    if ones_digit != 0:
        result += ONES_WORD[ones_digit]
    
    return result.strip()

test_cases = [0, 1, 13, 20, 42, 100, 115, 200, 742, 999]
for test_case in test_cases:
    print(number_to_words(test_case))   
````
