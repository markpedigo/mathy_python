## if statements

```python
age = 5

if age >= 18:
    print('You are an adult.')
else:
    print('You are not yet an adult.')

print('now we do the next part of the program.')
```

## if/ elif / else

```python
score = 84

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(grade)
```

## Combining conditions

```python
score = 75
attendance = 0.95

if score >= 80 and attendance >= 0.90:
    print('Honor roll')
else:
    print('Sorry, no honor roll for you.')
```

```python
ph = 7.7

if ph < 6.5 or ph > 8.5:
    print('pH out of safe range')
else:
    print('pH in safe range')
```

```python
is_failing = False

if not is_failing:
    print('On track')
```

```python
# short-circuit evaluation
x = 0

if (10 / x > 1) and x != 0:
    print('Big ratio')
else:
    print('Not a big ratio')
```

## A worked example

```python
temp_f = 62

if temp_f <= 32:
    category = 'freezing'
elif temp_f < 50:
    category = 'cold'
elif temp_f < 70:
    category = 'cool'
else:
    category = 'hot'

print(category)
```
