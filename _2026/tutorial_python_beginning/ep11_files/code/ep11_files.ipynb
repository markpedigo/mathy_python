## Opening and reading a file

```python
f = open('data.txt', 'r')  # file handle  # mode = --read
contents = f.read()
print(contents)
f.close()
```

```python
with open('data.txt', 'r') as f:
    contents = f.read()

print(contents)
```

## Reading line by line

```python
# readline() - reads one line at a time
with open('data.txt', 'r') as f:
    line1 = f.readline()
    line2 = f.readline()

print(line1)
print(line2)
```

```python
# readlines() - reads the entire file in one go
with open('data.txt', 'r') as f:
    for line in f:
        print(line.strip())
```

Method | Returns | Reads all at once?
-------|---------|-------------------|
readline() | one string | No
readlines() | list of strings | Yes
for line in f | one string per iteration | No

# Reading a csv

```python
with open('scores.csv', 'r') as f:
    lines = f.readlines()

header = lines[0].strip().split(',')
print(header)

for line in lines[1:]:
    fields = line.strip().split(',')
    name = fields[0]
    score = int(fields[1])
    grade = fields[2]
    print(f'{name}: {score} {grade}')
```
