6. Read `scores.csv` and compute the average score. Print it rounded to two decimal places. Do not hardcode the number of students.

````python
with open('data/scores.csv', 'r') as f:
    next(f)
    total = 0
    count = 0
    for line in f:
        fields = line.split(',')
        total += int(fields[1])
        count += 1
    average_score = total / count
    print(f'{average_score:.2f}')
````

7. Write a program that reads `scores.csv` and writes a new file `report.txt` containing a summary: total students, average score, highest score, lowest score, and the name of the top student.

````python
with open('data/scores.csv', 'r') as f:
    next(f)
    total = 0
    count = 0
    highest_score = -999
    top_student = ''
    lowest_score = 999
    for line in f:
        fields = line.split(',')
        name = fields[0]
        score = int(fields[1])
        total += score
        if score > highest_score:
            highest_score = score
            top_student = name
        if score < lowest_score:
            lowest_score = highest_score
        count += 1
    average_score = total / count

with open('data/report.txt', 'w') as f:
    f.write(f'{count}, {average_score}, {highest_score}, {lowest_score}, {top_student}')
````

9. Read `scores_for_pass_fail.csv` and write a new file `pass_fail.csv` with the same format, but add a fourth column 'result' that is 'Pass' if the score is 60 or above and 'Fail' otherwise. The header row should be `name,score,result`. Print the new file to verify.

````python
# Read scores_for_pass_fail.
with open('data/scores_for_pass_fail.csv', 'r') as f:
    next(f)
    lines = f.readlines()

# Parse
with open('data/pass_fail.csv', 'w') as f:
    f.write('name,score,result\n')
    for line in lines:
        line = line.strip()
        fields = line.split(',')
        score = int(fields[1])
        if score >= 60:
            fields.append('Pass')
        else:
            fields.append('Fail')
        f.write(",".join(fields) + "\n")

with open('data/pass_fail.csv', 'r') as f:
    contents = f.read()
    print(contents)
````

10. Write a robust file reader: a function `read_csv(filename)` that returns a list of dictionaries, one per row, using the header row as keys. For example, the first row of `scores.csv` should return `{'name': 'Alice', 'score': '92', 'grade': 'A'}`. Handle `FileNotFoundError` gracefully.

````python
def read_csv(filename):
    try:
        result = []
        with open(filename, 'r') as f:
            keys = f.readline().strip().split(',')
            for line in f:
                fields = line.strip().split(",")
                d = {}
                for i, key in enumerate(keys):
                    d[key] = fields[i]
                result.append(d)
        return result
    except:
        print(f'File not found: {filename}')
        return []

print(read_csv('data/scores.csv'))
````

11. **Word frequency from file.** Read `data.txt`, count the frequency of every word (ignoring case and punctuation), and write the results to `word_freq.txt` sorted from most to least frequent. Each line of the output file should look like: `the: 5`.

Use this line (or similar) to sort the dictionary. Don't worry about how it works for now. We'll cover lambdas in the intermediate course.
```python
freq = dict(sorted(freq.items(), key=lambda item: item[1], reverse=True))
```

````python
with open('data/data.txt', 'r') as f:
    frequencies = {}
    for line in f:
        words = line.split()
        for word in words:
            if word in frequencies:
                frequencies[word] += 1
            else:
                frequencies[word] = 1
        frequencies = dict(sorted(frequencies.items(), key=lambda item: item[1], reverse=True))

with open('data/word_freqency.txt', 'w') as f:
    for word, frequency in frequencies.items():
        f.write(word + ": " + str(frequency) + '\n')

with open('data/word_freqency.txt', 'r') as f:
    for line in f:
        print(line.strip())
````

12. **CSV to report.** Read `scores.csv` and write a formatted report to `report.txt` that looks like this:
```
Student Report
==============
Name       Score    Grade
Alice         92        A
Bob           75        C
Carol         88        B
Dave         100        A
Eve           84        B
==============
Average:   87.8
Top score: Dave (100)
```

````python
with open('data/scores.csv', 'r') as f_in:
    next(f_in)
    total = 0
    count = 0
    top_name = ''
    top_score = -999
    with open('data/report.txt', 'w') as f_out:
        f_out.write('Student Report\n')
        f_out.write('==============\n')
        f_out.write(f'{"Name":<10} {"Score":>7} {"Grade":>8}\n')
        for line in f_in:
            words = line.strip().split(",")
            name = words[0]
            score = int(words[1])
            grade = words[2]
            f_out.write(f'{name:<10} {score:>7} {grade:>8}\n')

            total += score
            count += 1

            if score > top_score:
                top_score = score
                top_name = name

        average = total / count
        f_out.write('==============\n')
        f_out.write(f'Average:   {average:.2f}\n')
        f_out.write(f'Top score: {top_name} ({top_score})')
````

13. **Index builder.** Read `data.txt` and build an inverted index: a dictionary mapping each unique word to the list of line numbers it appears on. Write the index to `index.txt`, one entry per line, sorted alphabetically. For example: `python: 1, 3, 5`.

````python
with open('data/data.txt', 'r') as f:
    word_index = {}
    for i, line in enumerate(f):
        words = line.strip().replace(".", "").split()
        print(i, words)
        for word in words:
            if word in word_index:
                word_index[word].append(i + 1)
            else:
                word_index[word] = [i + 1]
print(word_index)

with open('data/index.txt', 'w') as f:
    for word in word_index:
        f.write(word + ": " + str(word_index[word]) + "\n")
````
