## Common exception types

```python
# NameError - used aa name that is not defined
print(my_var)
```

```python
# TypeError - wrong type for the operation
'hello' + 5
```

```python
# Value Error - right type, invalid value
int('1.1')
```

```python
int(1.1)
```

```python
# ZeroDivisionError - divide by zero
10 / 0
```

```python
# IndexError - index is out of range
scores = [99, 92, 75]
scores[10]
```

```python
# KeyError - dictionary key does not exist
person = {
    'name': 'Mark'
}

person['age']
```

```python
# FileNotFoundError - file does not exist
open('missing.txt')
```

```python
# AttributeError - object does not have that method or attribute
my_tuple = (1, 2, 3)
my_tuple.append(4)
```

```python
# SyntaxError - Python cannot parse what you wrote
x === 3
```

Exception	| Cause	| Example
------------|-------|--------
NameError	|Used a name that isn't defined	| print(my_var) before assigning it
TypeError	|Wrong type for the operation	| 'hello' + 5
ValueError	|Right type, invalid value	| int('1.1')
ZeroDivisionError	|Divided by zero	| 10 / 0
IndexError	|Index out of range	| scores[10] on a 3-element list
KeyError	|Dictionary key doesn't exist	| person['age'] when 'age' isn't a key
FileNotFoundError	|File doesn't exist	| open('missing.txt')
AttributeError	|Object doesn't have that method	| .append() on a tuple
SyntaxError	|Python can't parse the code	 | x === 3


## `else` and `finally`

```python
try:
    result = divide(10, 2)
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(f'Result is {result}')
```

```python
def process_batch(scores):
    valid_count = 0
    try:
        for score in scores:
            if score < 0 or score > 100:
                raise ValueError(f'Score out of range: {score}')
            valid_count += 1
        print(f'All {valid_count} scores were valid.')
        return valid_count
    finally:
        print(f'Finished checking. Resetting counter for next batch.')
        valid_count = 0   # reset before this function call ends
```

```python
process_batch([85, 90, 78])
```

```python
process_batch([85, 200, 78])
```

```python
batches = [
    [85, 90, 78],
    [85, 200, 78],
    [60, 70, 80],
]

for batch in batches:
    try:
        result = process_batch(batch)
        print(f'  -> {result} valid scores.\n')
    except ValueError as e:
        print(f'  -> Batch failed: {e}\n')
```

# Robust File Reading

```python
filename = 'scores.csv'

try:
    with open(filename, 'r') as f:
        lines = f.readlines()
except FileNotFoundError:
    print(f'File not found: {filename}')
else:
    for line in lines[1:]:
        fields = line.strip().split(',')
        name = fields[0]
        score = int(fields[1])
        print(f'{name}: {score}')
```

## Raising exceptions

```python
def circle_area(radius):
    if radius < 0:
        raise ValueError(f'Radius cannot be negative: {radius}')
    return 3.14159 * radius ** 2
```

```python
circle_area(-5)
```

```python
circle_area(4)
```

## When to handle errors, or not

* Handle an exception when you can do something useful with it
* Let an exception go through when you cannot recover from it. e.g., Missing file
* Raise an exception when your function receives input that violates its contract.
