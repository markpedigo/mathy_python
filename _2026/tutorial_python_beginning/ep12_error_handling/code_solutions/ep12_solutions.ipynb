3. Write a function `safe_divide(a, b)` that returns `a / b` but catches `ZeroDivisionError` and returns `None` instead of crashing. Test it on `safe_divide(10, 2)` and `safe_divide(10, 0)`.

````python
def safe_divide(a, b):

    try:
        return a / b
    except ZeroDivisionError:
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

5. Write a function `safe_int(s)` that tries to convert a string to an integer and returns `None` if it fails. Test it on `'42'`, `'3.14'`, `'hello'`, and `'0'`.

````python
def safe_int(s):

    try:
        return int(s)
    except ValueError:
        return None

print(safe_int('42'))
print(safe_int('3.14'))
print(safe_int('hello'))
print(safe_int('0'))
````

6. Write a function `validate_triangle(a, b, c)` that raises a `ValueError` with a descriptive message if the sides don't form a valid triangle, and returns `True` otherwise.

````python
def validate_triangle(a, b, c):
    # To be valid, the sum of any two sides must be greater than the third side
    if (a + b > c) and (a + c > b) and (b + c > a):
        return True
    else:
        raise ValueError('Not a valid triangle.')
````

7. Write a function `safe_list_access(lst, index)` that returns the element at the given index, or a descriptive error message if the index is out of range. Test it on a list of five elements with valid and invalid indices.

````python
def safe_list_access(lst, index):
    try:
        return lst[index]
    except IndexError:
        print(f'Invalid index {index} for list {lst}.')

my_list = [1, 2, 3, 4, 5]

safe_list_access(my_list, 1)
safe_list_access(my_list, 10)
safe_list_access(my_list, -1)
safe_list_access(my_list, 10)
safe_list_access(my_list, 100)
````

8. Write a function `load_scores(filename)` that reads `scores.csv` and returns a list of `(name, score)` tuples. Use `finally` to print `'File operation complete.'` regardless of whether the read succeeded or failed.

````python
def load_scores(filename):
    try:
        with open(filename) as f:
            scores = []
            next(f)
            for lines in f:
                t = tuple(lines.strip().split())
                scores.append(t)
            return scores
    except FileNotFoundError:
        print(f'File not found: {filename}')
    finally:
        print('File operation complete.')

files = ['scores.csv', 'data/scores.csv', 'score.csv', 'scores.dat', 'scores.txt']
for file in files:
    print(file)
    scores = load_scores(file)
    if scores:
        print(scores)
    print('-' * 30)
````

9. Write a program that repeatedly asks the user for two numbers and divides them, catching `ZeroDivisionError` and `ValueError` separately. The program should keep running until the user types `'q'` for either number.

````python
while True:
    try:
        x = input('Enter the first number: ')
        if x == 'q':
            break

        y = input('Enter the second number: ')
        if y == 'q':
            break
        
        x = int(x)
        y = int(y)
        print(f'{x} / {y} = {x / y}')
        
    except ZeroDivisionError:
        print('Cannot divide by 0.')
    
    except ValueError:
        print('That is not a number.')
````

11. **Defensive calculator.** Write a function `calculate(expression)` that takes a string like `'10 / 2'` or `'5 + 3'` and evaluates it safely. Handle division by zero, invalid operators, non-numeric operands, and malformed expressions, each with a specific error message. Support `+`, `-`, `*`, and `/`. Do not use `eval()`.

````python
def calculate(expression):
    # supported operators
    operators = {'+', '-', '*', '/'}
    
    # parse the expression
    parts = expression.strip().split()
    
    # check for correct number of parts
    if len(parts) != 3:
        raise ValueError(f'Malformed expression: {expression}. Expected format: number operator number')
    
    left, operator, right = parts
    
    # validate operator
    if operator not in operators:
        raise ValueError(f'Invalid operator: {operator}. Supported operators: {", ".join(sorted(operators))}')
    
    # validate operands
    try:
        left = float(left)
    except ValueError:
        raise ValueError(f'Invalid left operand: {parts[0]}. Must be a number.')
    
    try:
        right = float(right)
    except ValueError:
        raise ValueError(f'Invalid right operand: {parts[2]}. Must be a number.')
    
    # perform calculation
    if operator == '+':
        return left + right
    elif operator == '-':
        return left - right
    elif operator == '*':
        return left * right
    elif operator == '/':
        if right == 0:
            raise ZeroDivisionError('Cannot divide by zero.')
        return left / right

# test cases
test_cases = [
    '10 / 2',        # valid
    '5 + 3',         # valid
    '7 * 4',         # valid
    '9 - 6',         # valid
    '10 / 0',        # ZeroDivisionError
    '5 + abc',       # invalid operand
    '5 $ 3',         # invalid operator
    '5 + 3 + 2',     # malformed
    '5',             # malformed
    '',              # malformed
]

for expr in test_cases:
    try:
        result = calculate(expr)
        print(f'{expr:>15} = {result}')
    except (ValueError, ZeroDivisionError) as e:
        print(f'{expr:>15} -> Error: {e}')
````
