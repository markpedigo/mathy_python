11. **Tip calculator.** A restaurant bill is \$47.35. Compute the tip and total for 15%, 18%, and 20% tip rates. Print each result on its own line, labeled clearly.

**Problem 13: Compound interest.** You invest \\$5,000 at 6% annual interest, compounded annually. The balance after $n$ years is given by:

$$B = P(1 + r)^n$$

where $P$ is the principal, $r$ is the annual interest rate, and $n$ is the number of years.

Compute the balance after 1, 5, 10, and 20 years.

The Rule of 72 is a mental math shortcut for estimating doubling time: divide 72 by the annual interest rate expressed as a percentage. At 6%, the Rule of 72 predicts doubling in $72 / 6 = 12$ years. Verify this estimate by computing balances year by year until the balance first exceeds $2 \times P$. How close was the Rule of 72?

````python
P = 5000
r = 0.06

B_1 = P * (1 + r)
B_5 = P * (1 + r)**5
B_10 = P * (1 + r)**10
B_20 = P * (1 + r)**20

print(f'${B_1:,.2f}')
print(f'${B_5:,.2f}')
print(f'${B_10:,.2f}')
print(f'${B_20:,.2f}')
print()

year = 1
while True:
    B = P * (1 + r)**year
    if B > 2 * P:
        print(year)
        break
    year += 1
````

14. **Gregorian calendar.** There are 365.2425 days in a year on average (accounting for leap years). Using only arithmetic operators, compute how many seconds are in a year, how many years are in a million seconds, and how many days old you are if you were born on January 1, 1969. Assume today is July 1, 2026.

````python
days_per_year = 365.2425
hours_per_day = 24
minutes_per_hour = 60
seconds_per_minute = 60

seconds_per_year = days_per_year * hours_per_day * minutes_per_hour * seconds_per_minute
print(f'Seconds in a year:      {seconds_per_year:>15,.0f}')

# years in a million seconds
million_seconds = 1000000
years_per_million = million_seconds / seconds_per_year
print(f'Years in 1M seconds:    {years_per_million:>15.4f}')

# days old if born January 1, 1969
# today is July 1, 2026
years_lived = 2026 - 1969
days_from_jan_to_jul = 31 + 28 + 31 + 30 + 31 + 30   # Jan through Jun
days_old = years_lived * days_per_year + days_from_jan_to_jul
print(f'Days old (born Jan 1, 1969): {days_old:>10,.0f}')
````

15. **Viète's formula.** The following product converges to $2/\pi$:

$$\frac{2}{\pi} = \frac{\sqrt{2}}{2} \cdot \frac{\sqrt{2+\sqrt{2}}}{2} \cdot \frac{\sqrt{2+\sqrt{2+\sqrt{2}}}}{2} \cdots$$

Using only `**` for square roots (recall $\sqrt{x} = x^{0.5}$), compute the first four terms of this product and estimate $\pi$. How accurate is your estimate?

````python
import math

# terms
t1 = 2 ** 0.5 / 2
t2 = (2 + 2 ** 0.5) ** 0.5 / 2
t3 = (2 + (2 + 2 ** 0.5) ** 0.5) ** 0.5 / 2
t4 = (2 + (2 + (2 + 2 ** 0.5) ** 0.5) ** 0.5) ** 0.5 / 2

# product of first four terms
product = t1 * t2 * t3 * t4

# estimate of pi
pi_estimate = 2 / product

print(f'Term 1:      {t1:.8f}')
print(f'Term 2:      {t2:.8f}')
print(f'Term 3:      {t3:.8f}')
print(f'Term 4:      {t4:.8f}')
print()

print(f'π estimate:  {pi_estimate:.8f}')
print(f'math.pi:     {math.pi:.8f}')
print(f'Difference:  {abs(pi_estimate - math.pi):.8f}')
````
