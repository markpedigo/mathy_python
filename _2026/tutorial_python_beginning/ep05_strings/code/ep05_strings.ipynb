## f-strings

```python
channel = 'Mathy Python'
episodes = 20
topic = 'strings'
```

```python
print(f'Welcome to {channel}. There might be {episodes} episodes. Today we cover {topic}.')
```

```python
print(f'area = {3.14 * 7 ** 2}')
```

```python
print(f'area = {3.1415926358 * 4.1 ** 2}')
```

```python
print(f'area = {3.1415926358 * 4.1 ** 2: .2f}')
```
