11. **Palindrome checker.** Write code that checks whether a string is a palindrome -- reads the same forwards and backwards -- ignoring spaces and case. Test it on `'racecar'`, `'A man a plan a canal Panama'`, and `'Mathy Python'`.

````python
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

print(is_palindrome('racecar'))
print(is_palindrome('A man a plan a canal Panama'))
print(is_palindrome('Mathy Python'))
````

12. **Caesar cipher, decrypted.** You receive this encrypted message:
```
Pdwkb Sbwkrq lv ixq
```
It was encrypted with a Caesar cipher using a shift of 3. To decrypt it, loop over each character, shift it back by 3 mod 26, and rebuild the string. Preserve spaces and punctuation. What does it say?

````python
def decrypt(message):
    ALPHABET_UPPER = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    ALPHABET_LOWER = 'abcdefghijklmnopqrstuvwxyz'
    
    result = ""
    for char in message:
        if char in ALPHABET_UPPER:
            pos = ALPHABET_UPPER.index(char)
            new_pos = (pos - 3) % 26
            new_char = ALPHABET_UPPER[new_pos]
        elif char in ALPHABET_LOWER:
            pos = ALPHABET_LOWER.index(char)
            new_pos = (pos - 3) % 26
            new_char = ALPHABET_LOWER[new_pos]
        else:
            new_char = char
        result += new_char
    return result

s = 'Pdwkb Sbwkrq lv ixq!'
print(decrypt(s))
````

13. **Word frequency.** Given any sentence string, count how many times each unique word appears, ignoring case and punctuation. Print the results. Test it on a sentence of your choice.

````python
sentence = 'How much wood would a woodchuck chuck if a woodchuck could chuck wood?'

sentence = sentence.lower().replace("?", "")  # more robust version would check for other punctuation

words = sentence.split()
words_unique = set(words)

words_dict = {}
for word in words_unique:
    
    if word not in words_dict.keys():
        word_count = words.count(word)
        words_dict[word] = word_count

for word in words_dict.keys():
    print(f'{word:10} {words_dict[word]}')
````
