11. **Triangular numbers.** The nth "triangular number" is $1 + 2 + 3 + ... + n$. Find the nth triangular number for n=1, n=10, and n=50.

````python
# Brute force with a loop
def triangular_number(n):
    t = 0
    for i in range(n + 1):
        t += i
    return t

print(triangular_number(1))
print(triangular_number(10))
print(triangular_number(50))
````

````python
# Using a formula
def triangular_number(n):
    return int(n * (n + 1) / 2)

print(triangular_number(1))
print(triangular_number(10))
print(triangular_number(50))
````

13. **Dictionary inversion.** Given this dictionary:
```python
grades = {'Alice': 'A', 'Bob': 'C', 'Carol': 'B', 'Dave': 'A'}
```
Build a new dictionary that maps grades to lists of students who received that grade. Expected result: `{'A': ['Alice', 'Dave'], 'C': ['Bob'], 'B': ['Carol']}`.

````python
grades = {'Alice': 'A', 'Bob': 'C', 'Carol': 'B', 'Dave': 'A'}

grades_reverse_lookup = {}
for key in grades.keys():
    value = grades[key]
    if value not in grades_reverse_lookup:
        grades_reverse_lookup[value] = [key]
    else:
        grades_reverse_lookup[value].append(key)
print(grades_reverse_lookup)
````

15. **The birthday problem.** In a group of 23 people, the probability that at least two share a birthday is just over 50%. The exact probability that at least 2 people share a birthday is:

$$P = 1 - \frac{365}{365} \times \frac{364}{365} \times \frac{363}{365} \times \cdots \times \frac{365-n+1}{365}$$

Compute this probability for $n = 23$. How does it compare to 50%? Then try $n = 30$ and $n = 50$.

````python
def bday_prob(n):
    P = 1
    for i in range(n):
        P *= (365 - i) / 365
    return (1 - P) * 100

print(f'n = 23: {bday_prob(23):.1f}')
print(f'n = 30: {bday_prob(30):.1f}')
print(f'n = 50: {bday_prob(50):.1f}')
````
