**Problem 13: Leap years.** A year is a leap year if it is divisible by 4, except century years (divisible by 100) are not leap years, unless they are also divisible by 400. So 1900 is not a leap year, but 2000 is.

Using only variables, `%`, and `==`, write expressions that correctly classify these four years: 1900, 1904, 2000, 2024.

No `if` statements allowed -- just `True`/`False` expressions.

**Problem 14: Seconds to time.** Given a number of seconds (say, 9999), compute and print how many hours, minutes, and remaining seconds that represents.

Your output should look like: `2 hours, 46 minutes, 39 seconds`.
