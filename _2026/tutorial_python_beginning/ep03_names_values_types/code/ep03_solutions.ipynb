Can you generalize it to a three-digit number?

````python
n = 99

# your code here
tens = n // 10
units = n % 10
total_digits = tens + units
print(f'{tens} + {units} = {total_digits}')
````

**Problem 12: Making change.** Write a program that takes a purchase amount and a payment amount (both in whole cents, e.g. 375 for $3.75), and prints how many dollars, quarters, dimes, nickels, and pennies to give back as change. For example, 41 cents change should be 1 quarter, 1 dime, 1 nickel, 1 penny.

Hint: `//` and `%` are your friends.

````python
payment = 500
purchase = 133

# your code here
change = payment - purchase
print(f'Your change is {change} cents.\n')

dollars = change // 100
change = change % 100

quarters = change // 25
change = change % 25

dimes = change // 10
change = change % 10

nickels = change // 5
pennies = change % 5


print(f'Dollars:  {dollars}')
print(f'Quarters: {quarters}')
print(f'Dimes:    {dimes}')
print(f'Nickels:  {nickels}')
print(f'Pennies:  {pennies}')
````

**Problem 14: Seconds to time.** Given a number of seconds (say, 9999), compute and print how many hours, minutes, and remaining seconds that represents.

Your output should look like: `2 hours, 46 minutes, 39 seconds`.

````python
total_seconds = 9999

# your code here
# Number of seconds in an hour = 60 sec / min * 60 min / hr = 3600 sec / hr
hours = total_seconds // 3600
remaining_secs = total_seconds % 3600
minutes = remaining_secs // 60
seconds = remaining_secs % 60

print(f'{hours:d} hours, {minutes:d} minutes, {seconds:d} seconds.')
````

**Problem 15: Compound interest surprise.** You deposit $1000 at 5% annual interest. Using only variables and arithmetic operators (no loops, no functions), compute the balance after 1 year, 5 years, and 10 years.

The formula is 

$$\text{balance} = \text{principal} \cdot (1 + \text{rate}) ^ {\text{years}}$$

How many years does it take to double? Try to get within one year using just exponentiation.

````python
principal = 1000
rate = 0.05

# your code here
balance_1y = principal * (1 + rate) ** 1
balance_5y = principal * (1 + rate) ** 5
balance_10y = principal * (1 + rate) ** 10

print(f'The balance after 1 year is: ${balance_1y:.2f}.')
print(f'The balance after 5 years is: ${balance_5y:.2f}.')
print(f'The balance after 10 years is: ${balance_10y:.2f}.')
````

````python
balance_15y = principal * (1 + rate) ** 15
print(f'The balance after 15 years is: ${balance_15y:.2f}.')
````
