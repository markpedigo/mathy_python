# Episode 2 - Generators and Iterators

## What Actually Happens in a `for` loop

```python
my_list = [1, 2, 3, 4, 5]
for x in my_list:
    print(x)
```

```python
it = iter(my_list)

print(next(it))   # 1
print(next(it))   # 2
print(next(it))   # 3
print(next(it))   # StopIteration
```

```python
r = range(10)
it = iter(r)
print(next(it))   # 0
print(next(it))   # 1
```

## Iterables vs Iterators

```python
# Loop over a list multiple times
print(my_list)
print(my_list)

# Loop over an iterator only once, then it's exhausted
print('-' * 10)
it = iter(my_list)
print(list(it))
print(list(it))
```

## Generator Functions

```python
def count_up(n):
    for i in range(1, n + 1):
        yield i

gen = count_up(5)
print(next(gen))
print(next(gen))
print(list(gen))
```

```python
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib = fibonacci()
for _ in range(10):
    print(next(fib))
```

## Generator Expressions Revisited

```python
# generator expression
gen = (n ** 2 for n in range(10))
list(gen)
```

```python
# equivalent generator function
def squares(n):
    for i in range(n):
        yield i ** 2

gen = squares(10)
list(gen)
```
