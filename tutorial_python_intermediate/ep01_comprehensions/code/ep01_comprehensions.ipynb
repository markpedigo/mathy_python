# From Loops to Comprehensions

```python
squares = []
for n in range(1, 11):
    squares.append(n ** 2)
print(squares)
```

```python
squares = [n ** 2 for n in range(1, 11)]
print(squares)
```

# Filtering with a condition

```python
evens = [n for n in range(1, 21) if n % 2 == 0]
print(evens)
```

```python
evens = []
for n in range(1, 21):
    if n % 2 == 0:
        evens.append(n)
print(evens)
```

# Dictionary and Set Comprehensions

```python
# dictionary
squares_dict = {n: n ** 2 for n in range(1, 11)}
print(squares_dict)
```

```python
# set
unique_remainders = {n % 7 for n in range(50)}
print(unique_remainders)
```

# Generator Expressions

```python
squares_gen = (n ** 2 for n in range(1, 11))
print(squares_gen)
print(list(squares_gen))
```

```python
total = sum(n ** 2 for n in range(1, 11))
print(total)   
```

# Nested comprehensions

```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
```

```python
flat = []
for row in matrix:
    for x in row:
        flat.append(x)
print(flat)
```

```python
flat = [x for row in matrix for x in row]
print(flat)
```

# Performance

```python
import timeit

comp_time = timeit.timeit(
    '[n ** 2 for n in range(10000)]',
    number=10000)

loop_time = timeit.timeit(stmt='''
result = []
for n in range(10000):
    result.append(n ** 2)
''', number=10000)

print(f'Comprehension: {comp_time:.3f}s')
print(f'Loop:          {loop_time:.3f}s')
```
